```python
# fashion_retail_analysis.ipynb

# %% [markdown]
# # Análisis de Fashion Retail Sales
# ## Análisis Exploratorio de Datos (EDA)

# %%
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# Configuración de estilo
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (12, 8)

# %%
# Cargar datos limpios
df = pd.read_csv('../data/processed/fashion_retail_sales_clean.csv')
df['date_purchase'] = pd.to_datetime(df['date_purchase'])

# Mostrar primeras filas
print("Primeras 5 filas del dataset:")
display(df.head())

# %%
# Información básica del dataset
print("="*50)
print("INFORMACIÓN BÁSICA DEL DATASET")
print("="*50)
print(f"Total de registros: {len(df):,}")
print(f"Período de datos: {df['date_purchase'].min()} to {df['date_purchase'].max()}")
print(f"Total de clientes únicos: {df['customer_id'].nunique():,}")
print(f"Total de productos únicos: {df['item_purchased'].nunique():,}")
print(f"Rango de precios: ${df['purchase_amount_usd'].min():.2f} - ${df['purchase_amount_usd'].max():.2f}")
print(f"Valores nulos por columna:")
print(df.isnull().sum())

# %%
# Estadísticas descriptivas
print("ESTADÍSTICAS DESCRIPTIVAS:")
print(df[['purchase_amount_usd', 'review_rating']].describe())

# %%
# Análisis de tendencias de ventas
print("\n" + "="*50)
print("ANÁLISIS DE TENDENCIAS DE VENTAS")
print("="*50)

# Ventas por mes
monthly_sales = df.groupby(['purchase_year', 'purchase_month']).agg({
    'purchase_amount_usd': ['sum', 'count', 'mean']
}).round(2)
monthly_sales.columns = ['total_sales', 'transaction_count', 'avg_transaction']

print("Ventas mensuales:")
display(monthly_sales)

# %%
# Visualización de tendencia de ventas mensuales
plt.figure(figsize=(14, 6))
monthly_trend = df.groupby(['purchase_year', 'purchase_month'])['purchase_amount_usd'].sum().reset_index()
monthly_trend['period'] = monthly_trend['purchase_year'].astype(str) + '-' + monthly_trend['purchase_month'].astype(str).str.zfill(2)

plt.subplot(1, 2, 1)
plt.plot(monthly_trend['period'], monthly_trend['purchase_amount_usd'], marker='o', linewidth=2, markersize=8)
plt.title('Monthly Sales Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Sales (USD)')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.bar(monthly_trend['period'], monthly_trend['purchase_amount_usd'])
plt.title('Monthly Sales', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Sales (USD)')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# %%
# Análisis de clientes
print("\n" + "="*50)
print("CLIENT ANALYSIS")
print("="*50)

customer_frequency = df.groupby('customer_id').agg({
    'purchase_amount_usd': ['count', 'sum', 'mean']
}).round(2)
customer_frequency.columns = ['purchase_count', 'total_spent', 'avg_spent']

print("Clients statistics:")
print(f"Customers with one purchase only: {(customer_frequency['purchase_count'] == 1).sum()}")
print(f"Recurrent customers: {(customer_frequency['purchase_count'] > 1).sum()}")
print(f"Customer that has spent the most: ${customer_frequency['total_spent'].max():.2f}")
print(f"Average expense of customers : ${customer_frequency['total_spent'].mean():.2f}")

# %%
# Distribución de gasto de clientes
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(customer_frequency['total_spent'], bins=30, alpha=0.7, edgecolor='black')
plt.title('Distribution of the Total expense per customer', fontsize=14, fontweight='bold')
plt.xlabel('Total expense (USD)')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.boxplot(customer_frequency['total_spent'])
plt.title('Boxplot of spending per customer', fontsize=14, fontweight='bold')
plt.ylabel('USD')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# %%
# Análisis de productos
print("\n" + "="*50)
print("PRODUCT ANALYSIS")
print("="*50)

# Productos más vendidos
top_products = df['item_purchased'].value_counts().head(10)
print("Top 10 most sold products:")
display(top_products)

# Productos con mayor revenue
product_revenue = df.groupby('item_purchased')['purchase_amount_usd'].agg(['sum', 'count', 'mean']).round(2)
product_revenue.columns = ['total_revenue', 'units_sold', 'avg_price']
top_revenue_products = product_revenue.sort_values('total_revenue', ascending=False).head(10)

print("\nTop 10 products per revenue:")
display(top_revenue_products)

# %%
# Visualización de análisis de productos
plt.figure(figsize=(15, 10))

# Top productos por unidades vendidas
plt.subplot(2, 2, 1)
top_products_df = top_products.reset_index()
plt.barh(top_products_df['item_purchased'], top_products_df['count'])
plt.title('Top 10 Most Sold Products', fontsize=14, fontweight='bold')
plt.xlabel('Sold Units')
plt.gca().invert_yaxis()

# Top productos por revenue
plt.subplot(2, 2, 2)
top_revenue_df = top_revenue_products.reset_index()
plt.barh(top_revenue_df['item_purchased'], top_revenue_df['total_revenue'])
plt.title('Top 10 Products per Revenue', fontsize=14, fontweight='bold')
plt.xlabel('Revenue (USD)')
plt.gca().invert_yaxis()

# Distribución de precios
plt.subplot(2, 2, 3)
plt.hist(df['purchase_amount_usd'], bins=30, alpha=0.7, edgecolor='black')
plt.title('Price Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Price (USD)')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Relación precio-unidades vendidas
plt.subplot(2, 2, 4)
plt.scatter(product_revenue['avg_price'], product_revenue['units_sold'], alpha=0.6)
plt.title('Relation: Price vs Sold Units', fontsize=14, fontweight='bold')
plt.xlabel('Average price (USD)')
plt.ylabel('Units Sold')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# %%
# Análisis de métodos de pago
print("\n" + "="*50)
print("PAYMENT METHODS ANALYSIS")
print("="*50)

payment_stats = df.groupby('payment_method').agg({
    'purchase_amount_usd': ['count', 'sum', 'mean'],
    'review_rating': 'mean'
}).round(2)
payment_stats.columns = ['transaction_count', 'total_amount', 'avg_amount', 'avg_rating']

print("Statistics by payment method:")
display(payment_stats)

# %%
# Visualización métodos de pago
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
payment_dist = df['payment_method'].value_counts()
plt.pie(payment_dist.values, labels=payment_dist.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribución de Métodos de Pago', fontsize=14, fontweight='bold')

plt.subplot(1, 3, 2)
plt.bar(payment_stats.index, payment_stats['avg_amount'])
plt.title('Ticket Promedio por Método de Pago', fontsize=14, fontweight='bold')
plt.xlabel('Método de Pago')
plt.ylabel('Ticket Promedio (USD)')
plt.xticks(rotation=45)

plt.subplot(1, 3, 3)
plt.bar(payment_stats.index, payment_stats['avg_rating'])
plt.title('Rating Promedio por Método de Pago', fontsize=14, fontweight='bold')
plt.xlabel('Método de Pago')
plt.ylabel('Rating Promedio')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

# %%
# Análisis de reseñas
print("\n" + "="*50)
print("ANÁLISIS DE RESEÑAS")
print("="*50)

review_stats = df['review_rating'].describe()
rating_distribution = df['review_rating'].value_counts().sort_index()

print("Estadísticas de reseñas:")
print(review_stats)

# Correlación entre rating y monto de compra
correlation = df[['review_rating', 'purchase_amount_usd']].corr().iloc[0,1]
print(f"\nCorrelación rating-monto: {correlation:.3f}")

# %%
# Visualización de reseñas
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
rating_distribution.plot(kind='bar')
plt.title('Distribución de Ratings', fontsize=14, fontweight='bold')
plt.xlabel('Rating')
plt.ylabel('Frecuencia')
plt.grid(True, alpha=0.3)

plt.subplot(1, 3, 2)
plt.scatter(df['purchase_amount_usd'], df['review_rating'], alpha=0.6)
plt.title('Relación Precio vs Rating', fontsize=14, fontweight='bold')
plt.xlabel('Precio (USD)')
plt.ylabel('Rating')
plt.grid(True, alpha=0.3)

plt.subplot(1, 3, 3)
rating_by_price = df.groupby(pd.cut(df['purchase_amount_usd'], bins=5))['review_rating'].mean()
rating_by_price.plot(kind='bar')
plt.title('Rating Promedio por Rango de Precio', fontsize=14, fontweight='bold')
plt.xlabel('Rango de Precio (USD)')
plt.ylabel('Rating Promedio')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

# %%
# Análisis estacional
print("\n" + "="*50)
print("ANÁLISIS ESTACIONAL")
print("="*50)

df['day_of_week'] = df['date_purchase'].dt.day_name()
df['month_name'] = df['date_purchase'].dt.month_name()

daily_sales = df.groupby('day_of_week')['purchase_amount_usd'].agg(['sum', 'count']).round(2)
monthly_sales_seasonal = df.groupby('month_name')['purchase_amount_usd'].agg(['sum', 'count']).round(2)

print("Ventas por día de la semana:")
display(daily_sales)
print("\nVentas por mes:")
display(monthly_sales_seasonal)

# %%
# Visualización estacional
plt.figure(figsize=(15, 6))

# Ventas por día de la semana
plt.subplot(1, 2, 1)
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
daily_sales_ordered = daily_sales.reindex(day_order)
daily_sales_ordered['sum'].plot(kind='bar')
plt.title('Ventas por Día de la Semana', fontsize=14, fontweight='bold')
plt.xlabel('Día de la Semana')
plt.ylabel('Ventas (USD)')
plt.grid(True, alpha=0.3)

# Ventas por mes
plt.subplot(1, 2, 2)
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 
               'July', 'August', 'September', 'October', 'November', 'December']
monthly_sales_ordered = monthly_sales_seasonal.reindex(month_order)
monthly_sales_ordered['sum'].plot(kind='bar')
plt.title('Ventas por Mes', fontsize=14, fontweight='bold')
plt.xlabel('Mes')
plt.ylabel('Ventas (USD)')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# %%
# Métricas clave finales
print("\n" + "="*50)
print("MÉTRICAS CLAVE DEL NEGOCIO")
print("="*50)

key_metrics = {
    'total_revenue': df['purchase_amount_usd'].sum(),
    'avg_transaction_value': df['purchase_amount_usd'].mean(),
    'total_customers': df['customer_id'].nunique(),
    'total_transactions': len(df),
    'avg_rating': df['review_rating'].mean(),
    'conversion_rate': len(df) / df['customer_id'].nunique(),
    'avg_spent_per_customer': df.groupby('customer_id')['purchase_amount_usd'].sum().mean()
}

for metric, value in key_metrics.items():
    if metric in ['total_revenue', 'avg_transaction_value', 'avg_spent_per_customer']:
        print(f"{metric}: ${value:,.2f}")
    else:
        print(f"{metric}: {value:,.2f}")

# %%
# Guardar análisis en CSV
monthly_sales.to_csv('monthly_sales_analysis.csv')
customer_frequency.to_csv('customer_analysis.csv')
product_revenue.to_csv('product_analysis.csv')
payment_stats.to_csv('payment_analysis.csv')

print("\nArchivos de análisis guardados:")
print("- monthly_sales_analysis.csv")
print("- customer_analysis.csv")
print("- product_analysis.csv")
print("- payment_analysis.csv")
```
